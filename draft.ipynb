```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Window length
N = 256

# Hann window
n = np.arange(N)
hann = 0.5 - 0.5 * np.cos(2 * np.pi * n / (N - 1))
```

```python
# Plot
plt.figure(figsize=(8, 4))
plt.plot(n, hann, linewidth=2)
plt.title(f"Hann Window (N={N})")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.xlim(0, N - 1)
plt.ylim(0, 1.05)
plt.show()
```
